```python
import matplotlib.pyplot as plt 
import numpy as np

x = np.arange(1,5,1)
print(x)
y = np.array([2,4,3,5])
plt.plot(x, y)
plt.show()
```

```python
y_mean  = y.mean()
y_means = np.ones(4)*y_mean
y_means

plt.plot(x, y, label="plot")
plt.plot(x, y_means, label="mean")
plt.show()
```

```python
y_over = y > y_means
y_overs = y[y_over]
x_overs = x[y_over]

plt.plot(x, y, label="plot", marker="o")
plt.scatter(x_overs, y_overs, linestyle="", color="red", label="over mean", zorder=3) # zorderは上下順

# plt.plot(x, y_means, label="mean")
plt.axhline(y_mean, linestyle="--", label="mean")
plt.legend(loc="upper left")
plt.show()
```

```python
rng = np.random.default_rng(1)
rand = rng.random(100)
print(rand)

y= rand
x = np.arange(1,len(y)+1)
print(x)
print(len(x), len(y))

y_mean = y.mean()

y_over = y[y>y_mean]
x_over = x[y>y_mean]

plt.plot(x,y)
plt.scatter(x_over, y_over, color="r", zorder=3)
plt.axhline(y_mean, linestyle="--")
plt.show()
```

```python
rng = np.random.default_rng(1)

num_data = 4
x = np.arange(1, 1+num_data, 1)
datas = [rng.random(100) for i in range(num_data)]

datas_mean = [data.mean() for data in datas]
datas_std = [data.std() for data in datas]

all_mean = np.mean(datas_mean)

fig, axes = plt.subplots(2,1)
fig = plt.figure(figsize=(10,5))
# ax1 = fig.add_subplot(1,1,1)
# ax2 = fig.add_subplot(2,1,1)
ax1 = axes[0]

ax1.axhline(all_mean, linestyle="--", label="mean:{:.4}".format(all_mean))
# ax.plot(datas_mean)
ax1.errorbar(x, datas_mean, yerr=datas_std, fmt="o", capsize=5)
# ax.set_xticks(datas_std)
ax1.set_title("title")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
ax1.legend()

# ax2.
plt.show()
```
